```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files
import numpy as np

# Upload one image
uploaded = files.upload()

img = cv2.imread(next(iter(uploaded)))

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect edges
edges = cv2.Canny(gray, 50, 150)

# Hough Line Transform
lines = cv2.HoughLinesP(
    edges,
    1,
    np.pi / 180,
    50,
    minLineLength=50,
    maxLineGap=10
)

# Draw detected lines
if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line
        cv2.line(img, (x1, y1), (x2, y2), (0, 0, 255), 2)

print("Detected Lines")
cv2_imshow(img)
```

```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload one image
uploaded = files.upload()

name = list(uploaded.keys())[0]

# Read image
img1 = cv2.imread(name)

# Create second image from the same image
img2 = cv2.resize(img1, None, fx=0.8, fy=0.8)

# Convert to grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# SIFT
sift = cv2.SIFT_create()

kp1, des1 = sift.detectAndCompute(gray1, None)
kp2, des2 = sift.detectAndCompute(gray2, None)

# Feature matching
bf = cv2.BFMatcher()

matches = bf.knnMatch(des1, des2, k=2)

good = []

for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good.append(m)

# Draw matches
result = cv2.drawMatches(
    img1, kp1,
    img2, kp2,
    good, None,
    flags=2
)

print("Good Matches:", len(good))

cv2_imshow(result)
```

```python
import cv2
import numpy as np
from google.colab import files

uploaded = files.upload()
img = cv2.imread(next(iter(uploaded)), 0)

# Resize image
img = cv2.resize(img, (100, 100))

# Convert image into data
data = img.astype(np.float32)

# PCA
mean, eigenvectors = cv2.PCACompute(data, mean=None, maxComponents=10)

# Project data
result = cv2.PCAProject(data, mean, eigenvectors)

print("Original Dimensions:", data.shape)
print("Reduced Dimensions:", result.shape)
```

```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload one image
uploaded = files.upload()

name = list(uploaded.keys())[0]

# Read image
img1 = cv2.imread(name)

# Create second image by resizing
img2 = cv2.resize(img1, None, fx=0.8, fy=0.8)

# 1. Grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# 2. Noise removal
gray1 = cv2.GaussianBlur(gray1, (5, 5), 0)
gray2 = cv2.GaussianBlur(gray2, (5, 5), 0)

# 3. SIFT
sift = cv2.SIFT_create()

kp1, des1 = sift.detectAndCompute(gray1, None)
kp2, des2 = sift.detectAndCompute(gray2, None)

# 4. Feature matching
bf = cv2.BFMatcher()

matches = bf.knnMatch(des1, des2, k=2)

good = []

for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good.append(m)

# 5. Display matched keypoints
result = cv2.drawMatches(
    img1, kp1,
    img2, kp2,
    good, None,
    flags=2
)

print("Number of matched keypoints:", len(good))

cv2_imshow(result)
```
